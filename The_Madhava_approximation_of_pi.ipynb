# The Madhava approximation of $\pi$

The Madhava series is a formula for calculating $\pi/4$:

$$ 
\frac{\pi}{4} =\sum_{n=0}^\infty \frac{(−1)^n}{2n+1}
$$

In this notebook, I am going to calculate an approximation of $\pi/4$ with various tolerance values between $10^{-1}$  and $10^{-7}$. and then compare their accuracy and time taken.

## Part A

```python
import numpy as np
```

```python
def Madhava_approximation_1(tolerance):
    """return the approximation of $\pi/4$ by using Madhava approximation, with given tolerance"""
    n=0
    newterm=((-1)**n)/(2*n+1)
    value=0

    while abs(newterm)>tolerance:
        value=value+newterm
        n=n+1
        newterm=((-1)**n)/(2*n+1)

    return value
```

```python
for i in range (1,8):
    diff=np.pi/4-Madhava_approximation_1(10**-i)
    print('The difference between pi/4 and tolerance 10^-'+str(i),'is:',diff)
```

Overall, our result is accurate to n decimal places for a tolerance $10^{-n}$ as differences only occur at n+1 decimal places. The reason might be as the next term is always smaller than the previous term, the new term that below tolerance is too small and can not change the decimal places above tolerance, therefore our result is accurate to $n$ decimal places for a tolerance of $10^{-n}$

## Part B

```python
def Madhava_approximation_2(tolerance):
    """return the approximation of $\pi/4$ by using Madhava approximation and number of terms, with given tolerance"""
    n=0
    newterm=((-1)**n)/(2*n+1)
    value=0

    while abs(newterm)>tolerance:
        value=value+newterm
        n=n+1
        newterm=((-1)**n)/(2*n+1)

    return value,n
```

```python
%%time

pi_over_4,num_terms=Madhava_approximation_2(10**-1)
diff=np.pi/4-pi_over_4
print('The difference between pi/4 and tolerance 10^-1 is:',diff,'. number of terms:',num_terms)
```
