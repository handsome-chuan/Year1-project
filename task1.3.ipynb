```python
#this code is used to calculate the final speed and time taken from v**2= u**2 + 2*a*s 
#(square of final velocity=sum of square of initial velocity and product of 2 and acceleration and displacement)
#debug by Yao Chengxi
#11.00 30/09/2025
####################################

#import sqrt

#wrong spelling of numpy, not 'numpty'
from numpy import sqrt

#initialise data

#I use acceleration to substitute g
#initial_velocity to substitute v0
#initial_displacement to substitute z0
#final_displacement to subsitute z
#final_velocity_square to substitute v2
#final_velocity to substitute speed
acceleration = -9.81
initial_velocity = 10
initial_displacement = 15
final_displacement = 0.0

#carry out the value of time taken and final velocity

#not 'G',it should be lower bound, which is 'g'
final_velocity_square = initial_velocity**2 + 2*acceleration*(final_displacement-initial_displacement)
#wrong spelling, not 'sqr', it should be 'sqrt'
final_velocity = sqrt(final_velocity_square)
if initial_velocity > 0:
#space and bracket are miss
    print("initial velocity is positive")
#colon is miss
else:
#brackets and quotation mark are miss
    print("initial velocity is less than or equal to zero")
time_taken = -(final_velocity + initial_velocity)/acceleration

#output the value of final velocity and time taken

#space before 'print'
#not meaningful output, ',' is missed
#symbol between time and taken is not minus symbol '-', it should be '_'
print('the final velocity is:',final_velocity,'m/s, the time taken is:',time_taken,'s')
```
