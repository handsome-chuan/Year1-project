# Part 1
In unit 2 i was asked to define a Python function to calculate the value of the fourth-order
Legendre polynomial $P_4(𝑥)$ for a given value (or array of values) of $x$. Now i am going yo plot a graph
of this function.
$$P_4(x) = \frac{1}{8} \left( 35 x^4 - 30 x^2 + 3 \right)$$

```python
def function(x):
    #calculates the function 1/8 * (35x^4 - 30x^2 + 3)
    data=(35*(x**4)-30*(x**2)+3)/8
    return data
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-1,1,1001) # 1001 points (1000 intervals) between -1 and 1
y = function(x)
```

```python
# Now for the plot itself
plt.figure()  # start a new figure
plt.plot(x,y) # plot the array "x" on the x axis and "y" on the y-axis
plt.xlabel("$x$") #label the x-axis
plt.ylabel('$P_4(𝑥)$') #label the y-axis
plt.title("A plot of $P_4(𝑥)$");  # Semicolon ; on last line suppresses unwanted text output
```

# Part2
This is an exercise in plotting data from a laboratory experiment.  A beam of X-rays is scattered from a crystal sample and the intensity of the scattered X-rays is measured at various angles.
In this part, i will plot both sets of data on a single graph, where the x-axis is twice the diffraction angle ($2\theta$, in degrees) and y-axis is the number of counts recorded per second

### Importing

Before plotting anything, we need to import the `pyplot` module from Matplotlib. We'll also need to import NumPy, as before.

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Reading data from a file
Then we need to read data from file so the program can execute it

```python
xdata1, ydata1 = np.loadtxt("LiF-30kV.csv",delimiter=',', unpack=True)   #load data from documents LiF-30kV.csv
xdata2, ydata2 = np.loadtxt("NaCl-30kV.csv",delimiter=',', unpack=True)  #load data from documents NaCl-30kV.csv
```

### Ploting a graph and Saving
Plot graph by using the data read before, and then set the legend,title,label and style use(colour)
Save the graph with name 'X-ray diffraction.png', so it is in png format

```python
plt.style.use('tableau-colorblind10')    # change style to make plots accessible for colour-blind users
plt.figure()                             # start a new figure
plt.title('X-ray diffraction')           # a title for the graph
plt.xlabel('2 theta/degrees')          # put a label on the x axis, including units
plt.ylabel('$N$/number of counts recorded per second')  #put a label on the y axis
plt.plot(xdata1,ydata1, 'o',label='LiF-30kV');  # plot number of counts vs $2\theta$ in LiF-30kV
plt.plot(xdata2,ydata2,'v',label='NaCl-30kV');  # plot number of counts vs $2\theta$ in NaCl-30kV

plt.legend(loc='best');                  # this will try and guess the best location for the legend...

plt.savefig('X-ray diffraction.png') # saves to a PNG file
```
